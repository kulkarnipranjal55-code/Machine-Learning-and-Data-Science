```python
#Assignment 1 : Tokenization & Money Extraction
import re
import pandas as pd
import numpy as np
```

```python
#Business Problem
'''Business Objective

Extract money transactions and currencies from text automatically.'''

#Business Constraints
'''Different currency symbols may exist.
Text format can vary.'''
```

```python
#Input text
text = "Tony gave two $ to Peter, Bruce gave 500 € to Steve"
```

```python
#tokenization
tokens = text.split()

print(tokens)
```

```python
#Extract money transaction using Regex
pattern = r'(\d+\s?[€$])'

money = re.findall(pattern, text)

print("Money Transactions:", money)
```

```python
#output
for i in money:
    print(i)
```

```python
#Interpretation
'''Interpretation
Regex identifies numbers followed by currency symbols.
Useful for finance and banking applications.'''

#Business Benefits
'''Automates transaction extraction.
Reduces manual effort.
Helps in financial text analytics.'''
```

```python
#Assignment 2 : Stemming and Lemmatization
!pip install nltk spacy

import nltk
import spacy

from nltk.stem import PorterStemmer

nltk.download('punkt')

nlp = spacy.load("en_core_web_sm")
```

```python
#Business Problem
'''Business Objective

Convert words into root/base form for better NLP processing.'''

#Business Constraints
'''Same word can appear in multiple forms.
Need meaningful normalization.'''
```

```python
#stemming e.g
ps = PorterStemmer()

doc1 = "Mando talked for 3 hours although talking isn't his thing"

words = doc1.split()

stemmed_words = [ps.stem(word) for word in words]

print(stemmed_words)
```

```python
#2nd doc stemming
doc2 = "eating eats eat ate adjustable rafting ability meeting better"

words = doc2.split()

stemmed = [ps.stem(word) for word in words]

print(stemmed)
```

```python
#stemming on list of words
lst_words = ['running', 'painting', 'walking', 'dressing',
             'likely', 'children', 'whom', 'good',
             'ate', 'fishing']

stemmed_words = []

for word in lst_words:
    stemmed_words.append(ps.stem(word))

print(stemmed_words)
```

```python
#lemmitization using spaCy
doc = nlp("running painting walking dressing likely children who good ate fishing")

lemmas = []

for token in doc:
    lemmas.append(token.lemma_)

print(lemmas)
```

```python
#full text stemming
text = """Latha is very multi talented girl.She is good at many skills like dancing,
running, singing, playing.She also likes eating Pav Bhagi.
she has a habit of fishing and swimming too.Besides all this,
she is wonderful at cooking too."""

words = text.split()

stemmed = [ps.stem(word) for word in words]

print(stemmed)
```

```python
#full text lemmetization
doc = nlp(text)

lemmas = []

for token in doc:
    lemmas.append(token.lemma_)

print(lemmas)
```

```python
#Interpretation
'''Interpretation
Stemming cuts words roughly.
Lemmatization gives meaningful root words.'''

#Business Benefits
'''Improves search engines.
Helps chatbot understanding.
Reduces duplicate vocabulary.'''
```

```python
#Assignment 3 : POS Tagging
import spacy
from collections import Counter

nlp = spacy.load("en_core_web_sm")
```

```python
#read text file
with open('/content/news_story.txt', 'r', encoding='utf-8') as file:
    text = file.read()

print(text[:500])
```

```python
#converting text into  NLP objects
doc = nlp(text)
```

```python
#extract NOUN Token
nouns = []

for token in doc:
    if token.pos_ == "NOUN":
        nouns.append(token.text)

print(nouns)
```

```python
#extract numbers
numbers = []

for token in doc:
    if token.pos_ == "NUM":
        numbers.append(token.text)

print(numbers)
```

```python
#count POS tags
pos_counts = Counter()

for token in doc:
    pos_counts[token.pos_] += 1

print(pos_counts)
```

```python
#visualization of POS tags
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))

plt.bar(pos_counts.keys(), pos_counts.values())

plt.xticks(rotation=45)

plt.title("POS Tag Counts")

plt.show()
```

```python
#Assignment 4 — Named Entity Recognition
import spacy

nlp = spacy.load("en_core_web_sm")
```

```python
text = """Kiran want to know the famous foods in each state of India.
So, he opened Google and search for this question.
Google showed that in Delhi it is Chaat,
in Gujarat it is Dal Dhokli,
in Tamilnadu it is Pongal,
in Andhrapradesh it is Biryani,
in Assam it is Papaya Khar,
in Bihar it is Litti Chowkha."""
```

```python
#extract GPE entities
doc = nlp(text)

locations = []

for ent in doc.ents:
    if ent.label_ == "GPE":
        locations.append(ent.text)

print("Geographical location Names:", locations)
```

```python
#birth date extraction
text = """Sachin Tendulkar was born on 24 April 1973,
Virat Kohli was born on 5 November 1988,
Dhoni was born on 7 July 1981
and finally Ricky Ponting was born on 19 December 1974."""
```

```python
#extract dates
doc = nlp(text)

dates = []

for ent in doc.ents:
    if ent.label_ == "DATE":
        dates.append(ent.text)

print("All Birth Dates:", dates)
```

```python
#Interpretation
'''Interpretation
NER extracts meaningful entities from text.
Identifies locations and dates automatically.'''

#Business Benefits
'''Useful in information extraction.
Helps search engines.
Supports intelligent document processing.'''
```

```python
#Assignment 5 — Bag of Words Movie Review Classification
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split

from sklearn.feature_extraction.text import CountVectorizer

from sklearn.naive_bayes import MultinomialNB

from sklearn.pipeline import Pipeline

from sklearn.metrics import classification_report, accuracy_score

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Business Problem
'''Business Objective

Classify movie reviews into positive or negative sentiments.'''

#Business Constraints
'''Reviews contain noisy text.
Model should predict sentiment accurately.'''
```

```python
df = pd.read_csv('/content/movie_reviews.csv')

df.head()
```

```python
df.info()

df.describe()
```

```python
#missing val
df.info()

df.describe()
```

```python
#EDA
sns.countplot(x=df['sentiment'])

plt.title("Sentiment Distribution")

plt.show()
```

```python
#1MBD
df['review_length'] = df['review'].apply(len)

df['review_length'].mean()

#2MBD
df['review_length'].var()

df['review_length'].std()

#3MBD
df['review_length'].skew()

#4MBD
df['review_length'].kurt()
```

```python
#train-test split
X_train, X_test, y_train, y_test = train_test_split(
    df['review'],
    df['sentiment'],
    test_size=0.2,
    random_state=42
)
```

```python
#building NLP pipeline
model = Pipeline([
    ('vectorizer', CountVectorizer()),
    ('classifier', MultinomialNB())
])
```

```python
#train model
model.fit(X_train, y_train)
```

```python
#prediction
y_pred = model.predict(X_test)
```

```python
#evaluation
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
```

```python
#classification report
print(classification_report(y_test, y_pred))
```

```python
#visualization
from collections import Counter

all_words = ' '.join(df['review'])

word_list = all_words.split()

common_words = Counter(word_list).most_common(10)

words = [i[0] for i in common_words]

counts = [i[1] for i in common_words]

plt.figure(figsize=(10,5))

plt.bar(words, counts)

plt.title("Top 10 Words")

plt.show()
```

```python
#Interpretation
'''Interpretation
Bag of Words converts text into numeric vectors.
Naive Bayes classifies sentiment efficiently.'''

#Business Benefits
'''Helps companies analyze customer opinions.
Automates sentiment analysis.
Improves decision making from reviews.'''
```
